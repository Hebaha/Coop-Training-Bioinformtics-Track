Q2- Does higher stress level correlate with poorer sleep quality and shorter sleep duration? 

```python

# Research Question:
# Does higher stress level correlate with poorer sleep quality
# and shorter sleep duration?


# Import required libraries
from scipy.stats import pearsonr
import matplotlib.pyplot as plt
import seaborn as sns


# Display summary statistics
print(df[['Stress_Level',
          'Quality_of_Sleep',
          'Sleep_Duration']].describe())

# Pearson Correlation Analysis


# Stress Level vs Quality of Sleep
r_quality, p_quality = pearsonr(
    df['Stress_Level'],
    df['Quality_of_Sleep']
)

# Stress Level vs Sleep Duration
r_duration, p_duration = pearsonr(
    df['Stress_Level'],
    df['Sleep_Duration']
)

print("="*60)
print("Pearson Correlation Results")
print("="*60)

print("\nStress Level vs Quality of Sleep")
print(f"Correlation coefficient (r): {r_quality:.3f}")
print(f"P-value: {p_quality:.5f}")

if p_quality < 0.05:
    print("Result: Significant correlation.")
else:
    print("Result: No significant correlation.")

print("\nStress Level vs Sleep Duration")
print(f"Correlation coefficient (r): {r_duration:.3f}")
print(f"P-value: {p_duration:.5f}")

if p_duration < 0.05:
    print("Result: Significant correlation.")
else:
    print("Result: No significant correlation.")
```

Q3- Which lifestyle factors are most strongly associated with sleep disorder in the dataset?

```python
# Add a new binary column for Sleep Disorder: 0 means no disorder, 1 means has a disorder
df['Sleep_Disorder_Binary'] = df['Sleep_Disorder'].apply(lambda x: 0 if str(x).strip().lower() == 'none' or pd.isna(x) else 1)

# map BMI categories to numeric values for regression analysis
bmi_mapping = {'Normal': 1, 'Normal Weight': 1, 'Overweight': 2, 'Obese': 3}
df['BMI_Numeric'] = df['BMI_Category'].map(bmi_mapping)

# Identify the dependent variable (Sleep Disorder) and independent variables (Lifestyle factors we want to study)
Y_q2 = df['Sleep_Disorder_Binary']

# the independent variables (lifestyle factors we want to study)
X_q2 = df[['Stress_Level', 'BMI_Numeric', 'Daily_Steps', 'Physical_Activity_Level']]

# add the constant (intercept) which is a mandatory step in regression analysis
X_q2 = sm.add_constant(X_q2)

# step 4: build the logistic regression model and fit it
# use the Logit function to build the logistic model
logit_model = sm.Logit(Y_q2, X_q2).fit()


# step 5: print the results and calculate odds ratios
print("=== Summary Results ===")
print(logit_model.summary())

print("\n=== Odds Ratios (Odds Ratios for Easier Interpretation) ===")
    
# calculate the odds ratios by taking the exponential of the coefficients
odds_ratios = np.exp(logit_model.params)
print(odds_ratios)
```

```python
params = logit_model.params
conf = logit_model.conf_int()

# Create a DataFrame for plotting the odds ratios and their confidence intervals
plot_df = pd.DataFrame({
    'Feature': params.index,
    'Odds_Ratio': np.exp(params),
    'Lower_CI': np.exp(conf[0]),
    'Upper_CI': np.exp(conf[1])
})

# Exclude the constant term (const) to focus the plot on the main variables only
plot_df = plot_df[plot_df['Feature'] != 'const']

# Improve the names of the variables for a more professional appearance in the plot
name_mapping = {
    'Stress_Level': 'Stress Level',
    'BMI_Numeric': 'BMI Category (Weight)',
    'Daily_Steps': 'Daily Steps',
    'Physical_Activity_Level': 'Physical Activity Level'
}
plot_df['Feature'] = plot_df['Feature'].map(name_mapping)

# Calculate the error for each variable programmatically (to display confidence interval lines)
plot_df['Error_Lower'] = plot_df['Odds_Ratio'] - plot_df['Lower_CI']
plot_df['Error_Upper'] = plot_df['Upper_CI'] - plot_df['Odds_Ratio']

# Set up the figure and style for the plot      
plt.figure(figsize=(10, 5))
sns.set_theme(style="whitegrid")

# Plot the points and error bars for the odds ratios
plt.errorbar(x=plot_df['Odds_Ratio'], y=plot_df['Feature'], 
             xerr=[plot_df['Error_Lower'], plot_df['Error_Upper']], 
             fmt='o', color='#e74c3c', markersize=10, elinewidth=3, capsize=5, label='Odds Ratio (OR)')

# Add a vertical line at 1 (the null effect line for odds ratios)
plt.axvline(x=1, color='black', linestyle='--', linewidth=2, label='Null Effect (OR = 1)')

# Enhance the aesthetics of the plot
plt.title('Lifestyle Predictors of Sleep Disorders\n(Forest Plot of Odds Ratios with 95% Confidence Intervals)', 
          fontsize=14, fontweight='bold', pad=15)
plt.xlabel('Odds Ratio (Values > 1 indicate Risk, < 1 indicate Protection)', fontsize=12, fontweight='bold')
plt.ylabel('Lifestyle & Physiological Factors', fontsize=12, fontweight='bold')
plt.legend(loc='lower right')

# Display the plot
plt.tight_layout()
plt.show()
```

Q4- Does resting heart rate differ between people with and without sleep disorders? 

```python


# Import required libraries
from scipy.stats import f_oneway
import seaborn as sns
import matplotlib.pyplot as plt



# Display descriptive statistics
print(df.groupby("Sleep_Disorder")["Heart_Rate"].describe())

# Create groups
healthy = df[df["Sleep_Disorder"] == "None"]["Heart_Rate"]

insomnia = df[df["Sleep_Disorder"] == "Insomnia"]["Heart_Rate"]

sleep_apnea = df[df["Sleep_Disorder"] == "Sleep Apnea"]["Heart_Rate"]

# Perform One-Way ANOVA
F_statistic, p_value = f_oneway(
    healthy,
    insomnia,
    sleep_apnea
)

print("="*60)
print("One-Way ANOVA Results")
print("="*60)

print(f"F-statistic : {F_statistic:.3f}")
print(f"P-value     : {p_value:.5f}")

if p_value < 0.05:
    print("Result: Significant difference among groups.")
else:
    print("Result: No significant difference among groups.")
```

```python

# Post-hoc Analysis (Tukey HSD)


from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey = pairwise_tukeyhsd(
    endog=df['Heart_Rate'],
    groups=df['Sleep_Disorder'],
    alpha=0.05
)

print(tukey)

tukey.plot_simultaneous(figsize=(8,5))
plt.title("Tukey HSD Multiple Comparisons")
plt.show()
```

Q5- Are individuals with insomnia more stressed than healthy individuals? 

```python

# Import required libraries
from scipy.stats import ttest_ind
import seaborn as sns
import matplotlib.pyplot as plt

#Null Hypothesis (H0):
# There is no significant difference in stress levels
# between individuals with insomnia and healthy individuals.
#
# Alternative Hypothesis (H1):
# Individuals with insomnia have higher stress levels
# than healthy individuals.


# Create the two groups
healthy = df[df["Sleep_Disorder"] == "None"]["Stress_Level"]

insomnia = df[df["Sleep_Disorder"] == "Insomnia"]["Stress_Level"]

# Display descriptive statistics
print("Healthy Group")
print(healthy.describe())

print("\nInsomnia Group")
print(insomnia.describe())

# Perform Welch's Independent Samples t-test
t_statistic, p_value = ttest_ind(
    insomnia,
    healthy,
    equal_var=False
)

print("="*60)
print("Independent Samples t-test Results")
print("="*60)

print(f"T-statistic : {t_statistic:.3f}")
print(f"P-value     : {p_value:.5f}")

if p_value < 0.05:
    print("Result: Significant difference between the two groups.")
else:
    print("Result: No significant difference between the two groups.")
```

```python
# Write the code of plotting
plt.figure(figsize=(7,5))

sns.boxplot(
    data=df[df["Sleep_Disorder"].isin(["None", "Insomnia"])],
    x="Sleep_Disorder",
    y="Stress_Level"
)

sns.stripplot(
    data=df[df["Sleep_Disorder"].isin(["None", "Insomnia"])],
    x="Sleep_Disorder",
    y="Stress_Level",
    color="black",
    alpha=0.4
)

plt.title("Stress Level in Healthy and Insomnia Groups")
plt.xlabel("Sleep Disorder")
plt.ylabel("Stress Level")

plt.show()
```

Q6- Are there a significant difference in sleep duration and sleep quality across different BMI categories? 

```python
# Sleep duration

# Import required library
from scipy.stats import f_oneway


# Null Hypothesis (H0):
# There is no significant difference in mean sleep duration
# across different BMI categories.
#
# Alternative Hypothesis (H1):
# At least one BMI category has a different mean sleep duration.


# Display descriptive statistics
print(df.groupby("BMI_Category")["Sleep_Duration"].describe())

# Create BMI groups
normal = df[df["BMI_Category"] == "Normal"]["Sleep_Duration"]
overweight = df[df["BMI_Category"] == "Overweight"]["Sleep_Duration"]
obese = df[df["BMI_Category"] == "Obese"]["Sleep_Duration"]

# Perform One-Way ANOVA
F_duration, p_duration = f_oneway(
    normal,
    overweight,
    obese
)

print("="*60)
print("One-Way ANOVA Results (Sleep Duration)")
print("="*60)

print(f"F-statistic : {F_duration:.3f}")
print(f"P-value     : {p_duration:.5f}")

if p_duration < 0.05:
    print("Result: Significant difference among BMI categories.")
else:
    print("Result: No significant difference among BMI categories.")
```

```python
# Sleep Quality


# Display descriptive statistics
print(df.groupby("BMI_Category")["Quality_of_Sleep"].describe())

# Create BMI groups
normal = df[df["BMI_Category"] == "Normal"]["Quality_of_Sleep"]
overweight = df[df["BMI_Category"] == "Overweight"]["Quality_of_Sleep"]
obese = df[df["BMI_Category"] == "Obese"]["Quality_of_Sleep"]

# Perform One-Way ANOVA
F_quality, p_quality = f_oneway(
    normal,
    overweight,
    obese
)

print("="*60)
print("One-Way ANOVA Results (Quality of Sleep)")
print("="*60)

print(f"F-statistic : {F_quality:.3f}")
print(f"P-value     : {p_quality:.5f}")

if p_quality < 0.05:
    print("Result: Significant difference among BMI categories.")
else:
    print("Result: No significant difference among BMI categories.")
```

```python
# plot 1
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="BMI_Category",
    y="Sleep_Duration"
)

sns.stripplot(
    data=df,
    x="BMI_Category",
    y="Sleep_Duration",
    color="black",
    alpha=0.4
)

plt.title("Sleep Duration Across BMI Categories")
plt.xlabel("BMI Category")
plt.ylabel("Sleep Duration (Hours)")

plt.show()


# plot 2
plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="BMI_Category",
    y="Quality_of_Sleep"
)

sns.stripplot(
    data=df,
    x="BMI_Category",
    y="Quality_of_Sleep",
    color="black",
    alpha=0.4
)

plt.title("Quality of Sleep Across BMI Categories")
plt.xlabel("BMI Category")
plt.ylabel("Quality of Sleep")

plt.show()
```

```python

# Post-hoc Analysis (Tukey HSD)
# Sleep Duration Across BMI Categories


# Perform Tukey HSD to determine which BMI categories
# differ significantly in mean sleep duration.

from statsmodels.stats.multicomp import pairwise_tukeyhsd

tukey_duration = pairwise_tukeyhsd(
    endog=df["Sleep_Duration"],
    groups=df["BMI_Category"],
    alpha=0.05
)

print(tukey_duration)

# Plot Tukey HSD confidence intervals
tukey_duration.plot_simultaneous(figsize=(8,5))

plt.title("Tukey HSD Comparison for Sleep Duration")
plt.xlabel("Mean Difference")

plt.show()
```

```python

# Post-hoc Analysis (Tukey HSD)
# Quality of Sleep Across BMI Categories

# Perform Tukey HSD to determine which BMI categories
# differ significantly in mean sleep quality.

tukey_quality = pairwise_tukeyhsd(
    endog=df["Quality_of_Sleep"],
    groups=df["BMI_Category"],
    alpha=0.05
)

print(tukey_quality)

# Plot Tukey HSD confidence intervals
tukey_quality.plot_simultaneous(figsize=(8,5))

plt.title("Tukey HSD Comparison for Quality of Sleep")
plt.xlabel("Mean Difference")

plt.show()




## The obese group contained a relatively small number of participants compared with the other BMI categories. 
#This imbalance may have reduced the statistical power to detect differences involving the obese group and should be considered
#when interpreting the results.

```

Q7- Is short sleep duration associated with an increased prevalence of elevated blood pressure in the dataset?

```python
df.columns = df.columns.str.strip().str.replace(' ', '_')

# Identify the dependent variable (Sleep Duration) and independent variables (Systolic and Diastolic Blood Pressure)
sleep_dur = df['Sleep_Duration']
systolic = df['Systolic_BP']
diastolic = df['Diastolic_BP']

# Perform Pearson correlation tests to assess the relationship between sleep duration and blood pressure
r_sys, p_sys = stats.pearsonr(sleep_dur, systolic)

r_dia, p_dia = stats.pearsonr(sleep_dur, diastolic)

# Print the results in a structured format
print("=== Q7: Pearson Correlation Results ===")
print(f"1. Sleep Duration vs Systolic BP:")
print(f"   - Correlation Coefficient (r): {r_sys:.4f}")
print(f"   - p-value: {p_sys:.4e}")

print(f"\n2. Sleep Duration vs Diastolic BP:")
print(f"   - Correlation Coefficient (r): {r_dia:.4f}")
print(f"   - p-value: {p_dia:.4e}")
```
